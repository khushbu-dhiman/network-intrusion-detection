```python
import pandas as pd
import os

base_path = r"C:\Users\hp\OneDrive\Desktop\ml_model_training"




df1 = pd.read_parquet(os.path.join(base_path, "Benign-Monday-no-metadata.parquet"))
df2 = pd.read_parquet(os.path.join(base_path, "DoS-Wednesday-no-metadata.parquet"))
df3 = pd.read_parquet(os.path.join(base_path, "Portscan-Friday-no-metadata.parquet"))
df4 = pd.read_parquet(os.path.join(base_path, "DDoS-Friday-no-metadata.parquet"))
df5 = pd.read_parquet(os.path.join(base_path,"Infiltration-Thursday-no-metadata.parquet"))
df6 = pd.read_parquet(os.path.join(base_path, "Bruteforce-Tuesday-no-metadata.parquet"))
df7 = pd.read_parquet(os.path.join(base_path, "Botnet-Friday-no-metadata.parquet"))

df = pd.concat([df1, df2, df3,df4,df5,df6,df7], ignore_index=True)

print(df.shape)
print(df['Label'].value_counts())
```

```python
dos_attacks = [
    'DoS Hulk',
    'DoS GoldenEye',
    'DoS slowloris',
    'DoS Slowhttptest'
]

df['Label'] = df['Label'].replace(dos_attacks, 'DoS')
```

```python
brute_force = ['FTP-Patator', 'SSH-Patator']

df['Label'] = df['Label'].replace(brute_force, 'BruteForce')
```

```python
df = df[~df['Label'].isin(['Heartbleed', 'Infiltration'])]
```
